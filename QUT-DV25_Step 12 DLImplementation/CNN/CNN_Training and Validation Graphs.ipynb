```python
import pandas as pd
import matplotlib.pyplot as plt
results_file_path = 'Result.csv'
results_df = pd.read_csv(results_file_path)

# Check the structure of the dataset
results_df.head()
```

```python
best_epoch = results_df['Validation Accuracy'].idxmax()
best_epoch_value = results_df.loc[best_epoch, 'Epoch']
best_validation_acc = results_df.loc[best_epoch, 'Validation Accuracy']
plt.figure(figsize=(10, 5))
plt.plot(results_df['Epoch'], results_df['Train Accuracy'], label='Train Accuracy')
plt.plot(results_df['Epoch'], results_df['Validation Accuracy'], label='Validation Accuracy')
plt.axvline(x=best_epoch_value, color='red', linestyle='--', label=f'Model Checkpoint ({best_epoch_value})')
plt.scatter(best_epoch_value, best_validation_acc, color='red', zorder=5)
plt.text(best_epoch_value, best_validation_acc, f"{best_validation_acc:.4f}", color="red", fontsize=10, ha="left")
plt.title('Training and Validation Accuracy over Epochs')
plt.xlabel('Number of Epoch')
plt.ylabel('Accuracy')
plt.legend(loc='lower right') 
plt.grid(True)
plt.ylim(0.45, 1.05)
plt.savefig('Training and Validation Accuracy over Epochs', dpi=300, bbox_inches="tight")
plt.show()
```

```python
best_epoch = results_df.loc[results_df['Epoch'] == 72].index[0]
best_epoch_value = 72
best_validation_loss = results_df.loc[best_epoch, 'Validation Loss']

plt.figure(figsize=(10, 5))
plt.plot(results_df['Epoch'], results_df['Train Loss'], label='Train Loss')
plt.plot(results_df['Epoch'], results_df['Validation Loss'], label='Validation Loss')
plt.axvline(x=best_epoch_value, color='red', linestyle='--', label=f'Model Checkpoint ({best_epoch_value})')
plt.scatter(best_epoch_value, best_validation_loss, color='red', zorder=5)
plt.text(best_epoch_value, best_validation_loss, f"{best_validation_loss:.4f}", color="red", fontsize=10, ha="left")
plt.title('Training and Validation Loss over Epochs')
plt.xlabel('Number of Epoch')
plt.ylabel('Loss')
plt.legend(loc='upper left') 
plt.grid(True)
plt.ylim(-0.05, 1.05)
plt.savefig('Training and Validation Loss over Epochs', dpi=300, bbox_inches="tight")
plt.show()
```
